# PLP Data Analysis Assignment 4: Agent Data Cleanup
This notebook runs the data cleaning and quality engineering pipeline to process raw transaction telemetry and mapping registries.

```python
import pandas as pd
from cleaning_pipeline import clean_agent_data

# Read raw transaction and dimension files
txns = pd.read_csv('agent_transactions_raw.csv')
agents = pd.read_csv('agents_raw.csv')

# Run the robust pipeline
clean_df, audit_log = clean_agent_data(txns, agents)

# Display pipeline metrics and audit logs
print(f'Before Row Count: {len(txns)}')
print(f'After Row Count: {len(clean_df)}')
display(audit_log)
```

```python
# Generate County Aggregations
county_summary = clean_df.groupby('county', dropna=False).agg(
    total_transactions=('txn_id', 'count'),
    total_volume=('amount', 'sum'),
    mean_amount=('amount', 'mean')
).reset_index()
display(county_summary)
```
